Knowledge Representation using Production Rules

```python
rules = [
    {
        "conditions": ["problem solving", "algorithms", "math"],
        "recommendation": "Artificial Intelligence",
        "confidence": 0.9
    },
    {
        "conditions": ["programming", "databases", "logic"],
        "recommendation": "Software Engineering",
        "confidence": 0.85
    },
    {
        "conditions": ["networks", "security", "linux"],
        "recommendation": "Cybersecurity",
        "confidence": 0.8
    },
    {
        "conditions": ["design", "creativity", "user experience"],
        "recommendation": "UI/UX Design",
        "confidence": 0.75
    },
    {
        "conditions": ["data", "statistics", "analysis"],
        "recommendation": "Data Science",
        "confidence": 0.88
    }
]
print("Saved Successfuly")
```

```python
def advise(interests):
    matches = []

    for rule in rules:
        required = rule["conditions"]

        # Forward chaining: check if all conditions are met
        if all(item in interests for item in required):
            matches.append({
                "path": rule["recommendation"],
                "confidence": rule["confidence"]
            })

    if not matches:
        return "Recommendation: No suitable course path found."

    # Resolve conflict using highest confidence
    best_choice = max(matches, key=lambda x: x["confidence"])
    confidence_percent = int(best_choice["confidence"] * 100)

    return f"Recommended Path: {best_choice['path']} (Confidence: {confidence_percent}%)"
```

```python
user_input = input("Enter your interests/skills separated by commas: ")
interests = [i.strip().lower() for i in user_input.split(",")]

result = advise(interests)
print(result)
```
