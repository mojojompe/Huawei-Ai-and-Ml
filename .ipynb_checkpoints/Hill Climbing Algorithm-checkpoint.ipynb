Hill Climbing Algorithm

```python
import random
print("Imported")
```

```python
def f(x):
    return -(x - 2)**2 + 4
def hill_climbing(start, step=0.1, max_iter=100):
    current = start
    for _ in range(max_iter):
        neighbors = [current + step, current - step]
        best_neighbor = max(neighbors, key=f)
        if f(best_neighbor) <= f(current):
            break
        current = best_neighbor
    return current, f(current)
```

```python
solution, value = hill_climbing(start=random.uniform(-5, 5))
print("Best x:", solution)
print("Maximum value:", value)
```
